```python
#nomor 1
bulan=(input("bulan yang ingin menginap "))
lama=int(input("berapa lama menginap "))
tipe=input("tipe kamar anda yang inginkan (studio/apartemen) ")

tarif_studio=0
tarif_apartemen=0

if bulan=="mei"or bulan=="oktober":
  tarif_studio =500000
  tarif_apartemen=650000
elif bulan=="juni"or bulan=="september":
  tarif_studio =752000
  tarif_apartemen=687000
elif bulan=="juli" or bulan=="agustus":
  tarif_studio=760000
  tarif_apartemen=770000
else :
  print("tidak tersedia ruangan di bulan ini")

total_tarif=0

if tipe=="studio":
  total_tarif=tarif_studio*lama
  if lama>3:
    diskon=total_tarif*0.15
    total_tarif-=diskon

elif tipe=="apartemen":
  total_tarif=tarif_apartemen*lama
  if lama>3:
    diskon=total_tarif*0.10
    total_tarif-=diskon

else:
  print("tipe kamar tidak valid")

if total_tarif>0:
  print(f"total tarif untuk menginap selama {lama} hari dibulan {bulan} dengan tipe kamar {tipe} adalah {total_tarif}")
```

```python
print("soal a")
data_pengiriman = ( ("Laptop", 2.5, "Surabaya", 35000),
                  ("Printer", 6.0, "Jakarta", 55000),
                  ("Monitor", 4.2, "Bandung", 45000),
                  ("Keyboard", 1.2, "Surabaya", 30000),
                  ("Mouse", 0.5, "Jakarta", 25000) )
dari_surabaya=[item for item in data_pengiriman if item[2]=="Surabaya"]
for item in dari_surabaya:
  print(item)

print("\nsoal b")
total_berat_jakarta=sum(item[1]for item in data_pengiriman if item[2]=="Jakarta")
print(f"total berat jakarta: {total_berat_jakarta}")

#soal c
print("\nsoal c")
data_terbaru=[]
for nama,berat,kota,biaya in data_pengiriman:
  if kota=="Bandung":
    biaya=(int(biaya*1.1))
    data_terbaru.append((nama,berat,kota,biaya))
print(data_terbaru)
```

```python
import math

a=4
b=10
c=4

root1=(-b + math.sqrt(b**2 - 4*a*c)) / (2*a)
root2=(-b - math.sqrt(b**2 - 4*a*c)) / (2*a)
diff=[]
for x in range(1,11):
  f_prime=2*a*x+b
  diff.append(f_prime)
print(f"root 1: {root1}")
print(f"root 2: {root2}")
print(f"f'(x): {diff}")
```

```python
# Memulihkan fungsi input bawaan Python yang tertimpa variabel
try:
  del input
except NameError:
  pass

nama_barang=str(input("nama barang yang ingin dibeli: "))
jumlah_awal=float(input("jumlah awal barang: "))
satuan_awal=str(input("satuan awal barang(pcs/lusin/kodi): "))
satuan_tujuan=str(input("satuan yang diinginkan(pcs/lusin/kodi): "))

error=False

if satuan_awal=="pcs":
  total_pcs=jumlah_awal
elif satuan_awal=="lusin":
  total_pcs=jumlah_awal*12
elif satuan_awal=="kodi":
  total_pcs=jumlah_awal*20
else:
  error=True

if not error:
  if satuan_tujuan=="pcs":
    total_tujuan=total_pcs
  elif satuan_tujuan=="lusin":
    total_tujuan=total_pcs/12
  elif satuan_tujuan=="kodi":
    total_tujuan=total_pcs/20
  else:
    error=True

if error:
  print("error satuan tidak dapat dikenali")
else:
  print(f"barang yang dibeli: {nama_barang}")
  print(f"jumlah awal: {jumlah_awal} {satuan_awal}")
  print(f"total dalam pcs:  {total_pcs}")
  print(f"jumlah yang diinginkan: {total_tujuan} {satuan_tujuan}")
```
